## 5. Numerical Analysis with NumPy
Using `.to_numpy()`, we perform direct array-based numerical analysis on grades (`G1`, `G2`, `G3`), `absences`, `studytime`, and `failures`:
- Central tendency & spread (`mean`, `median`, `std`, `var`, `IQR`)
- Percentiles (`25th`, `50th`, `75th`, `90th`)
- Pass rate (`G3 >= 10` on the 0–20 grading scale) and zero-grade counts
- Pearson correlation matrix using `np.corrcoef`

```python
def analyze_grades_numpy(df: pd.DataFrame, label: str) -> None:
    """Compute NumPy summary statistics for final grades (G3) and related metrics."""
    g1 = df["G1"].to_numpy(dtype=float)
    g2 = df["G2"].to_numpy(dtype=float)
    g3 = df["G3"].to_numpy(dtype=float)
    absences = df["absences"].to_numpy(dtype=float)

    q25, q50, q75, q90 = np.percentile(g3, [25, 50, 75, 90])
    iqr = q75 - q25
    pass_rate = np.mean(g3 >= 10) * 100
    zero_count = int(np.sum(g3 == 0))
    grade_change = g3 - g1

    print(f"=== NumPy Analysis: {label} (n={len(g3)}) ===")
    print(f"  G3 Mean              : {np.mean(g3):.2f}")
    print(f"  G3 Median            : {np.median(g3):.2f}")
    print(f"  G3 Std Dev           : {np.std(g3):.2f}")
    print(f"  G3 Variance          : {np.var(g3):.2f}")
    print(f"  G3 Min / Max         : {np.min(g3):.0f} / {np.max(g3):.0f}")
    print(f"  G3 Percentiles       : Q25={q25:.1f}, Q50={q50:.1f}, Q75={q75:.1f}, Q90={q90:.1f} (IQR={iqr:.1f})")
    print(f"  Pass Rate (G3 >= 10) : {pass_rate:.2f}%")
    print(f"  Students with G3 = 0 : {zero_count}")
    print(f"  Mean G1 -> G3 Change : {np.mean(grade_change):+.2f}")
    print(f"  Absences Mean / Max  : {np.mean(absences):.2f} / {np.max(absences):.0f}\n")


analyze_grades_numpy(df_math, "Mathematics")
analyze_grades_numpy(df_por, "Portuguese")
```

```python
# Compute Pearson correlation matrix with NumPy for key numerical features
key_features = ["studytime", "failures", "absences", "G1", "G2", "G3"]

for name, df in [("Mathematics", df_math), ("Portuguese", df_por)]:
    matrix = df[key_features].to_numpy(dtype=float)
    # rowvar=False treats each column as a variable
    corr_matrix = np.corrcoef(matrix, rowvar=False)
    corr_df = pd.DataFrame(corr_matrix, index=key_features, columns=key_features).round(3)
    print(f"=== NumPy Correlation Matrix ({name}) ===")
    display(corr_df)
```

## 6. Save Processed Datasets Separately
We keep `data/raw/` completely untouched and save clean, standard comma-separated CSV files in `data/processed/`:
1. `student_math_clean.csv` — Math dataset with a `subject` label and helper columns (`passed`, `avg_prior_grade`)
2. `student_por_clean.csv` — Portuguese dataset with the same helper columns
3. `student_combined_clean.csv` — Both subjects stacked together (`subject` identifies Math vs Portuguese; note that per `data/raw/student.txt`, 382 students appear in both courses)

We add two simple derived columns for future exploration:
- `avg_prior_grade`: average of `G1` and `G2`
- `passed`: `1` if `G3 >= 10`, else `0`

```python
def prepare_clean_copy(df: pd.DataFrame, subject_name: str) -> pd.DataFrame:
    """Return a copy of the dataset with standard helper columns added."""
    clean_df = df.copy()
    clean_df["subject"] = subject_name
    clean_df["avg_prior_grade"] = ((clean_df["G1"] + clean_df["G2"]) / 2.0).round(2)
    clean_df["passed"] = (clean_df["G3"] >= 10).astype(int)
    return clean_df


math_clean = prepare_clean_copy(df_math, "Math")
por_clean = prepare_clean_copy(df_por, "Portuguese")
combined_clean = pd.concat([math_clean, por_clean], ignore_index=True)

math_out = PROCESSED_DIR / "student_math_clean.csv"
por_out = PROCESSED_DIR / "student_por_clean.csv"
combined_out = PROCESSED_DIR / "student_combined_clean.csv"

math_clean.to_csv(math_out, index=False)
por_clean.to_csv(por_out, index=False)
combined_clean.to_csv(combined_out, index=False)

# Verify saved files can be read back cleanly
for path in [math_out, por_out, combined_out]:
    check_df = pd.read_csv(path)
    print(f"Verified {path.name}: shape={check_df.shape}, missing={int(check_df.isnull().sum().sum())}")
```
